## 1 GC content

$\dfrac{1}{L}\sum_{i=1}^{L} \mathbb{1}[\,b_i \in \{G, C\}\,]$

<br>

| symbol | meaning | python |  |
| --- | --- | --- | --- |
| $L$ | length of the sequence | len(sequence) |
| $b_i$ | base at current position | base |
| $\mathbb{1}[\,b_i \in \{G, C\}\,]$ | add 1 if the base is "G" or "C", otherwise add 0 | if base == "G" or "C": |
| $\sum_{i=1}^{L}$ | repeat that for every base and add them up | for base in sequence: gc_count = gc_count + 1 |
| $\dfrac{1}{L}$ | divide by L | gc_count / len(sequence) |
<br>

So GC content can be calculated by counting the number of Gs and Cs and then dividing by the total number of bases.

```python
def gc_content(sequence):  #we define the function
  """Calculates the fraction of GC in a sequence"""
  gc_count = 0 # start a counter at 0
  for base in sequence:
    if base == "G" or base == "C": # count this position if it is G or C
      gc_count = gc_count + 1 # if it is G or C, the counter is increased by 1
  return gc_count / len(sequence)
```

```python
# we test the counter
sequence = "AACGTCG"
print(gc_content(sequence))
```

## Hamming distance  

How many positions two equal-length sequences differ at. In general, how many positions two strings. So in words: we add one to a count when two sequences differ at one position, otherwise 0.

$D = \sum_{i=1}^{L} \mathbb{1}[\,a_1 \neq b_i\,]$  

<br>

| variable | meaning | python |
| --- | --- | --- |
| $D$ | Hamming distance | def_hamming_distance |
| $L$ | Length of the sequence | len(sequence) |
| $i$ | current position in sequence | i |
| $a_i$ | base at position i in the first sequence | seq1[i]|
| $b_i$ | base at position i in the second sequence | seq1[i] |
| $\neq$ | is not equal to | seq1[i] != seq2[i]|
| $1[...]$ | indicator: gives 1 when the condition is true and 0 when false | |
| $\sum{}$ | add up all 1s and 0s | hd_count = hd_count + 1|
| i = 1 to L | check every position from first character to the last | for i in range(len(seq1))|

<br>

So the overall logic of this is:  
1) start differences at 0
2) check position 0,1,2,...,n-1
3) if the two characters are different: add 1
4) after all the positions are checked (n-1): return the total

```python
def hamming_distance(seq1, seq2): # create a function named hamming_distance with inputs seq1 and seq2
  """How many positions differ between two strings of the same length"""
  hd_count = 0 # we start the counter
  for i in range(len(seq1)): # go through every position in the sequence, one at a time
  # i stores that current position
  # range(n) generates the index numbers 0,1,...,n-1, which we can then use to access positions in an ordered sequence like a list or string
    if seq1[i] != seq2[i]: # the comparison
    # seqx[i] means: get the character from seqx at position i
    # != means: is not equal to
      hd_count = hd_count + 1 # take the old value of hd_count and add 1 and save the new value back into hd_count
  return hd_count
```

```python
seq1 = "AACGTCG"
seq2 = "AACTTAG"
print(hamming_distance(seq1, seq2))
```

## Dot product
How much two vectors of numbers agree (How much the two lists of numbers point in the same overall direction) - and a weighted sum, which is a shape we need again later in the course.  

More context:  
1) large positive dot product --> strong agreement
2) large negative dot product --> strong disagreement
3) around zero --> little directional agreement.  

<br>


In plain english:
Multiply the values at each position in sequence *a* by the value at the same position in sequence *b*, then add all those products together.


<br>

$\sum_{i=1}^{N} a_i b_i$

<br>

| variable | meaning |
| --- | --- |
| $i$ | current position |  
| $a_i$ | value at position $i$ in sequence a |
| $b_i$ | value at position $i$ in sequence b |
| $a_1*b_1$ | multiply the two values at the same position |
| $N$ | total number of positions |
| $\sum_{i=1}^{N}$ | sum every position starting at 1 up to k |

<br>
Soooo: multiply matching values first, then add all of those products to get one final result.

```python
def running_total(seq1,seq2):
  """Running total"""
  total = 0
  # start the accumulator at 0

  totals = []
  # create an empty list
  # This list will store the value of the running total after every position

  for i in range(len(seq1)):
    # Go through every index position i in seq1

    product = seq1[i] * seq2[i]
    # take the valule from position i at seq1 and multiply by value at position i in seq2
    # store the result in product

    total = total + product
    # add the current product to the accumulated total.

    totals.append(total) # take the list totals and use its append action to add total
    # add the current value of total to the list totals.
    # this saves the total so far before moving to the next position.
    # totals is the list
    # append is a built-in function that lists know to do
    # total is the value we want to add

  return totals
  # after the loop is finished, return the entire list of running totals
```

```python
def final_total(seq1,seq2):
  """Final dot product"""
  total = 0
  # start a running total at 0, it will store the accumulated sum of the products

  for i in range(len(seq1)):
    # len(seq1) gives the number of values in seq1
    # range(len(seq1)) generates the index positions up to n-1
    # i is the current position in the sequence

    product = seq1[i] * seq2[i]
    # seq1[i] gets value from seq1 at position i
    # seq2[i] gets value from seq3 at position i
    # the two values are multiplied together and stored in the variable product

    total = total + product
    # The current product is added to the total accumulated so far
    # The new value replaces the old value of total

  return total
  # after all positions have been processed, return the final accumulated total
```

```python
seq1 = (1,2,3,4)
seq2 = (5,6,7,8)
print("dot product=",final_total(seq1,seq2))
print("running total=",running_total(seq1,seq2))
```

## Geometric mean

An average for quantities that multiply, like fold changes. So it gives us the typical multiplicative factor across several values. E.g. we ask: what if the average fold change? If all the different multiplicative changes were replaced by one constant factor, what factor would represent them overall?

<br>

$\left(\prod_{i=1}^{N} x_i\right)^{1/N}$

<br>

| | |
| --- | --- |
| $x_i$ | the value at position i |
| $i$ | the current position |
| $N$ | the total number of values |
| $\prod$ | multiply all the values together |
| $\prod_{i=1}^{N} x_i$ | multiply every value from the first through the Nth |
| $\dfrac{1}{N}$ | take the Nth root |



```python
def geometric_mean(values): #creates a function called geomatric means that takes one sequence of numbers called values
  product = 1 # starts the running product at 1 (multiplying by 0 obv wouldn't work)

  for value in values: # loops through every number in the sequence
    product = product * value # multiplies the product-so-far by the current value

  return product ** (1 / len(values))
  # len(values) gives N, the number of values
  # 1 / len(values) gives 1/N
  # ** means "raise to power"
```

```python
values = [1,2,3,4,5,6]
print(geometric_mean(values))
```

## Harmonic Mean
An average for rates. Take the reciprocal of each value, add those together, then divide the number of values by that sum. We use reciprocals because rates like speed are inverted when calculating time; averaging those reciprocals correctly accounts for how much time is spent at each rate.

<br>

$N \Big/ \displaystyle\sum_{i=1}^{N} \frac{1}{x_i}$

<br>

| | |
| --- | --- |
| $x_i$ | the value at position $i$ |
| $i$ | the current position |
| $N$ | the total number of values |
| $\frac{1}{x_i}$ | the reciprocal of the current value |
| $\sum_{i=1}^{N} \frac{1}{x_i}$ | add all the reciprocals |
| $N/$that sum | divide the number of values by the reciprocal sum |

```python
def harmonic_mean(values):
  reciprocal_sum = 0 # we create the variable and start the count at 0

  for i in values: # go through every number in values
    reciprocal_sum = reciprocal_sum + 1 / i
    # 1/i means: take the reciprocal of the current value

  return len(values) / reciprocal_sum
  # len(values) means: how many values are there
```

```python
print(harmonic_mean(values))
```

## Binomial coefficient
How many ways to choose *k* things out of *N*. Order does not matter here! For example, let's say we have 4 students (A,B,C,D) and we want to choose 2 of them.  

The possible groups are:  
AB, AC, AD, BC, BD, CD.

There are 6 possible groups, so:  

$\dbinom{4}{2} = 6$  

AB and BA are considered the same choise. This is what "order does not matter" refers to.  

So in an even plainer english:  

The binomial coefficient gives the number of different ways to select *k* items from a total of *N* items when the order of the selected items does not matter.

<br>

$\dbinom{N}{k} = \dfrac{N!}{k!\,(N-k)!}$

<br>



```python
# this is a bit more difficult, we first need to define the factorial

def factorial(n): # create a function that calculates n!
  product = 1 # since this is a product, we start at 1

  for value in range(1, n+1): # we loop through the intergers from 1 up to and including n
    product = product * value

  return product
```

```python
# now we apply that for the formula
def binomial_coefficient(N, k):
  numerator = factorial(N) # this calculates N!

  denominator = factorial(k) * factorial(N-k) # this calculates k!(N-k)!

  return numerator / denominator #this is just the fraction
```

```python
N = 8
k = 2

print(binomial_coefficient(N,k))
```
