**1. Classify a sequencing read.** `phages.fasta` holds T4, T7 and M13, and section 1 already
measures the base frequencies that separate them. Treat each phage as a hypothesis, build its
four base probabilities with `p_base`, and take the likelihood of a 40-base read under a phage
to be $\prod_{i} p_H(b_i)$, the product of the probabilities that phage gives to the bases we
saw. That product is an accumulator like section 1's, starting at 1.0 and multiplying rather
than adding. Feeding the three likelihoods to `posterior` classifies the read. A classifier
that multiplies the bases as though they were drawn separately is called naive Bayes, and
section 3 measured that they aren't, so its likelihoods come out more extreme than they should
be. It's used in production anyway, in spam filtering and in read binning, because the ranking
of the hypotheses often survives an assumption the individual numbers don't.

**2. Then run a prior against it.** Take a read from M13 and classify it twice, once with the
three phages equally likely and once with a prep that is 80% T7, 15% T4 and 5% M13. Bases 1431

on identical likelihoods: the read favours M13 by a factor of 4 and the prior favours T7 by a
factor of 16. Reads long enough to settle it break the arithmetic instead, and a 400-base
product falls under $10^{-308}$ and comes back as 0.0 for all three, which is a real limit of
this implementation and has a standard fix with a name.

**3. Condition on the previous base.** Replace $\prod_i p_H(b_i)$ with

## 1. Calculate how common A, C, G, and T are in each phage genome.


```python
# First, I will make a probability formula by counting
def base_freq(base,seq):
  """How often does each base occur in a sequence?"""
  count = 0
  for letter in seq:
    if letter == base:
      count = count + 1
  return count / len(seq)
```

## 5. Choose priors for T4, T7, and M13. If we have no prior information, use equal priors of 1/3 each.

Okay, so for the first classification, we need a **flat prior**. We have no reason to think T4, T7 or M13 is more likely than others.

So here, the prior is an assumption we choose because we have no other information.

Important: the prior comes from what we know about the possible sources before the examining read.

```python
# we make the priors
priors = {
    "T4": 1/3,
    "T7": 1/3,
    "M13": 1/3,
}
```

```python
# now we calculate the three likelihoods and store them
likelihoods = {}

for name, genome in zip(NAMES, GENOMES):
  likelihoods[name] = read_likelihood(read, genome)

print(likelihoods)
```

## 6. Combine each likelihood with its prior by multiplying them

Now we have two separate pieces of information.  
priors: before looking at the read, all are equally likely
posteriors: after looking at the read, how well does the read fit each phage?

<br>

The next step is to combine each likelihood with its prior, so:
for each phage, multiply "how well the read fits" by "how likely that phage was beforehand"

```python
def make_joint(likelihoods, priors):
# make a function that needs two inputs
# the likelihood and prior dictionaries

  joint = {} # creates an empty dictionary, where we will store the new values

  for name in likelihoods: # go through all phage names
    joint[name] = likelihoods[name] * priors[name]
    # multiplies each prior with each likelihood

  return joint
```

```python
joint = make_joint(likelihoods, priors)
print(joint)
```

## 7. Get evidence
This was only the stage 1 of the posterior calculation. After this, we now have to add the three values together and divide each one by that total.  

In Bayes, the total is called **evidence** or **marginal**.  

We need to:
1. Start the total at 0
2. go through T4, T7, M13
3. Add each joint value to the total
4. Return the total.

This will give us the total amount of probability mass across all three possible explanations.

```python
def total_evidence(joint):
  evidence = 0

  for name in joint:
    evidence = evidence + joint[name]

  return evidence
```

```python
evidence = total_evidence(joint)
print(evidence)
```

## 8. Turn the joint values into final probabilities

1. Create an empty dictionary for the final probabilities
2. Go through T4, T7, M13
3. For each phage, divide its joint value by the total evidence
4. store that result
5. return all three probabilities

```python
def make_posterior(joint, evidence):
  posterior = {}

  for name in joint:
    posterior[name] = joint[name] / evidence

  return posterior
```

```python
posterior = make_posterior(joint, evidence)
print(posterior)
```
